Tempurature Catagories:


*   Below 15 Celsius: Cold
*   15 to 25 Celsius (Inclusive): Warm
*   Above 25 Celsius: Hot




```python
# The list of temperatures provided
temperatures = [10, 16, 23, 28, 14, 22, 30]

# Loop through each temperature and apply the specific criteria
for t in temperatures:
  if t < 15:
    label = "Cold"
  elif t <= 25: # Less than or equal to 25 handles the "inclusive" rule
    label = "Warm"
  else:
    label = "Hot"

  # Formula to convert Celsius to Fahrenheit
  fahrenheit = (t * 9/5) + 32

  # Print containing the ° symbol and both temperature formats
  # round(farhenheit,1) keeps decimal clean
  print(f"{t}°C ({round(fahrenheit,1)}°F) is {label}")
```
