```python
# ============================
# Step 1: Wages with f-strings
# ===================================
# Create variables with my own values
# ===================================
name = "Marc"
hourly_wage = 22.5
hours = 30

# Compute weekly pay
weekly_pay = hourly_wage * hours

# print using an f-string
print(f"{name} earns {weekly_pay:.1f} per week.")
```

```python
# ===========================
# Step 2: Count the warm days
# ========================================
# Create the provided list of temperatures
# ========================================
temps = [61, 65, 72, 58, 70, 74, 68]

#Loop and condition pattern to count days strictly above 65
warm_day_count = 0
for t in temps:
  if t > 65: # Strictly greater than 65 (not >=)
    warm_day_count += 1

print(warm_day_count)
```

```python
# =====================================
# Step 3: Write a temperature converter
# =====================================
# Define the conversion function
# ==============================
def to_celsius(f):
  return (f - 32) * 5/9

# Call it on the first tempurature in the list (temps[0] is 61)
converted_temp = to_celsius(temps[0])

# Round the result to 1 decimal place and print
print(round(converted_temp, 1))
```

```python
# ==========================
# Step 4: A movie dictionary
# =================================
# Create a dictionary for one movie
# =================================
movie = {"title": "Toy Story 3", "year": 2010, "rating": 9.9}

# Print eack key and value usinf a loop
for key, value in movie.items():
  print(f"{key}: {value}")
```
