```python
# Function using a for loop
def multiply_with_for(num1, num2):
  total = 0
  for _ in range(num2):
    total += num1
  return total

# Function using a while loop
def multiply_with_while(num1, num2):
  total = 0
  counter = 0
  while counter < num2:
    total += num1
    counter += 1 # Updates counter so that loop doesn't run forever
  return total

# Call both function to prove they match the actual product (7 * 6 = 42)
print("For loop approach (7 * 6):", multiply_with_for(7, 6))
print("While loop approach (7 * 6):", multiply_with_while(7, 6))
print("Actual product (7 * 6):", 7 * 6)
```
